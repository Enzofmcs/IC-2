```python
# prompt: traduza o ultimo gráfico plotado e coloque em ordem decrescente

import matplotlib.pyplot as plt
# Dicionário com as traduções corretas
classes = {4: ('nv', 'Nevos melanocíticos'),
           6: ('mel', 'Melanoma'),
           2: ('bkl', 'Lesões benignas semelhantes a ceratose'),
           1: ('bcc', 'Carcinoma basocelular'),
           5: ('vasc', 'Granulomas piogênicos e hemorragia'),
           0: ('akiec', 'Ceratoses actínicas e carcinoma intraepitelial'),
           3: ('df', 'Dermatofibroma')}

# Crie um dicionário para traduzir os diagnósticos
traducao_diagnosticos = {v[0]: v[1] for v in classes.values()}

# Crie uma nova coluna 'dx_pt' com as traduções
ham['dx_pt'] = ham['dx'].map(traducao_diagnosticos)

# Traduza os tipos de diagnóstico
traducao_tipos = {
    'histo': 'Histopatológico',
    'consensus': 'Consenso',
    'confocal': 'Microscopia Confocal',
    'follow_up': 'Acompanhamento'
}
ham['dx_type_pt'] = ham['dx_type'].map(traducao_tipos)

# Conte o número de cada diagnóstico em português, ordenados em ordem decrescente
contagem_diagnosticos_pt = ham['dx_pt'].value_counts()

# Crie um gráfico de barras agrupadas com as traduções e ordenação
plt.figure(figsize=(12, 6))
sns.countplot(x='dx_type_pt', hue='dx_pt', data=ham, order=ham['dx_type_pt'].value_counts().index)
plt.xlabel('Tipo de Diagnóstico')
plt.ylabel('Contagem')
plt.title('Distribuição das Classes de Diagnóstico por Tipo de Diagnóstico')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Diagnóstico', loc='upper right')
plt.show()
```

```python
# Agrupa os dados por gênero e localização, contando as ocorrências


grouped_data = ham.groupby(['sex', 'localizacao_pt'])['dx'].count().reset_index(name='count')

# Cria o gráfico de barras agrupadas
plt.figure(figsize=(12, 6))  # Ajusta o tamanho da figura
sns.barplot(x='localizacao_pt', y='count', hue='sex', data=grouped_data)
plt.xlabel('Localização da Doença')
plt.ylabel('Número de Casos')
plt.title('Distribuição da Localização da Doença por Gênero')
plt.xticks(rotation=45, ha='right')  # Rotaciona os rótulos do eixo x para melhor legibilidade
plt.legend(title='Gênero')
plt.tight_layout()  # Ajusta o layout para evitar sobreposição de elementos
plt.show()
```

```python
# prompt: faça um gráfico parecido com o anterior, mas relacionando genero com doenças

import matplotlib.pyplot as plt
# Agrupa os dados por gênero e doença, contando as ocorrências
grouped_data = ham.groupby(['sex', 'dx_pt'])['dx'].count().reset_index(name='count')

# Cria o gráfico de barras agrupadas
plt.figure(figsize=(12, 6))  # Ajusta o tamanho da figura
sns.barplot(x='dx_pt', y='count', hue='sex', data=grouped_data)
plt.xlabel('Doença')
plt.ylabel('Número de Casos')
plt.title('Distribuição das Doenças por Gênero')
plt.xticks(rotation=45, ha='right')  # Rotaciona os rótulos do eixo x para melhor legibilidade
plt.legend(title='Gênero')
plt.tight_layout()  # Ajusta o layout para evitar sobreposição de elementos
plt.show()
```
