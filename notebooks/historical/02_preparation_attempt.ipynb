```python
mel_ds = ds[ds['dx'] == 'mel']
mel_ds.shape
```

```python
for index, row in mel_ds.iterrows():
  id = row['image_id']
  image_path_part1 = os.path.join('HAM10000_images_part_1', f'{id}.jpg')
  image_path_part2 = os.path.join('HAM10000_images_part_2', f'{id}.jpg')
  destination_path = os.path.join('mel', f'{id}.jpg')
  try:
    shutil.copy(image_path_part1, destination_path)
  except FileNotFoundError:
    shutil.copy(image_path_part2, destination_path)
```
