# Nova execução de treinamento

Este roteiro depende das imagens e de PyTorch. Não foi executado durante a curadoria. Instale `python -m pip install -e ".[train]"` na raiz do repositório. Os comandos abaixo são exemplos de terminal e não executam automaticamente neste caderno.


## 1. Dados

Guarde os metadados em `data/raw/HAM10000_metadata.csv` e as imagens originais em `data/raw/images/`. Não copie imagens aumentadas para splits diferentes.

```sh
python -m ic2.data --metadata data/raw/HAM10000_metadata.csv --image-root data/raw/images --output data/processed/split.csv
```


## 2. Treinamento

```sh
python -m ic2.train --manifest data/processed/split.csv --image-root data/raw/images --architecture efficientnet_b0 --output runs/efficientnet_001
```

Troque architecture por `mobilenet_v3_large` ou `vgg16` e use outra pasta de saída. Reutilize o mesmo manifesto para comparações.


## 3. Avaliação

A execução salva `best.pt`, `history.json`, `metrics.json` e `test_predictions.csv`. Compare os hashes de manifesto e a política de seleção antes de comparar resultados. Mantenha os dados externos e pesos fora do Git.

O módulo novo usa head binário de uma saída e backbone congelado, portanto não carrega automaticamente os checkpoints antigos de duas saídas.
