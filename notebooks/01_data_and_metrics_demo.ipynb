# Separação e métricas: exemplo sintético

Sem imagens, sem downloads e sem treinamento. Os identificadores e probabilidades abaixo são inventados para explicar o fluxo.


```python
from ic2.data import split_metadata, validate_manifest
from ic2.metrics import binary_metrics, select_threshold
rows = [dict(image_id=f"{label}_{group}_{image}", lesion_id=f"{label}_{group}", dx=label)
        for label in ("mel", "nv") for group in range(10) for image in range(2)]
manifest = split_metadata(rows)
validate_manifest(manifest)
counts = {s: sum(r["split"] == s for r in manifest) for s in ("train", "val", "test")}
print(counts)
```

```python
for lesion in {r["lesion_id"] for r in manifest}:
    assert len({r["split"] for r in manifest if r["lesion_id"] == lesion}) == 1
threshold = select_threshold([0, 0, 1, 1], [0.1, 0.2, 0.3, 0.4])
report = binary_metrics([0, 1], [0.25, 0.35], threshold)
assert report["accuracy"] == 1
print(report)
```

A métrica perfeita deste exemplo é uma consequência dos números inventados. Não representa resultado da pesquisa. Em dados reais, a independência por lesão e a seleção do limiar na validação devem ser verificadas antes do teste.
